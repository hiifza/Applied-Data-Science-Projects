## 2 — What Is Data Visualization?

> *"The greatest value of a picture is when it forces us to notice what we never expected to see."*
> — John W. Tukey

**Data visualization** is the discipline of encoding data values into visual properties — position, colour, size, shape — so that patterns, relationships, and anomalies become immediately perceptible to the human eye. It is not decoration; it is the primary interface between raw numbers and human understanding.

### Why Visualize Before Modelling?

| Without Visualization | With Visualization |
|---|---|
| Outliers and distributions invisible | Immediately spotted in a boxplot |
| Non-linear relationships missed | Revealed in scatter plots |
| Class imbalance overlooked | Clear in a count/bar chart |
| Feature interactions hidden | Surface in heatmaps and facet grids |
| Summary statistics mislead (Anscombe's Quartet) | Plots expose the true data shape |

### Why Dashboards Matter

A dashboard is not a collection of charts — it is a **curated visual argument**. Each chart answers one question; the sequence of charts builds a narrative that a stakeholder can follow without statistical training. In professional data analysis, the ability to construct a clear visual story is often more valuable than the ability to run a model.

### What This Dashboard Answers

1. Who were the passengers aboard the Titanic?
2. What demographic and economic patterns existed?
3. Which factors most strongly determined survival?
4. How did class, gender, age, and family interact?
5. What can the data tell a stakeholder in under five minutes?

## 4 — Data Cleaning

```python
# ─────────────────────────────────────────────────────────────────────────────
# CLEANING PIPELINE
# ─────────────────────────────────────────────────────────────────────────────
print('BEFORE CLEANING')
print(f'  Shape         : {df.shape}')
print(f'  Total missing : {df.isnull().sum().sum()}')
print()

# ── Step 1: Age — median imputation ───────────────────────────────────────────
# Median is preferred over mean when the distribution may be skewed.
# With 177 missing values, median is robust and avoids outlier influence.
age_median = df['Age'].median()
df['Age']  = df['Age'].fillna(age_median)
print(f'[1] Age  → filled {177} NaNs with median = {age_median}')

# ── Step 2: Embarked — mode imputation ───────────────────────────────────────
# Only 2 rows affected. Mode ('S') is the clear majority (72%+).
emb_mode    = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(emb_mode)
print(f'[2] Embarked → filled 2 NaNs with mode = "{emb_mode}" (Southampton)')

# ── Step 3: Cabin — drop column ───────────────────────────────────────────────
# 77.1% missing. Imputation would be speculation, not recovery.
df.drop(columns=['Cabin'], inplace=True)
print('[3] Cabin → column dropped (77.1% missing — imputation not feasible)')

print()
print('AFTER CLEANING')
print(f'  Shape         : {df.shape}')
print(f'  Total missing : {df.isnull().sum().sum()}')
print('  ✓ Dataset is clean.')
```

## 5 — Feature Engineering

```python
# ─────────────────────────────────────────────────────────────────────────────
# ENGINEER FOUR FEATURES
# ─────────────────────────────────────────────────────────────────────────────

# FamilySize = SibSp + Parch  (PDF definition — does NOT add 1)
# Captures total connected family members aboard (excluding self)
df['FamilySize'] = df['SibSp'] + df['Parch']

# IsAlone = 1 if no family aboard
df['IsAlone'] = (df['FamilySize'] == 0).astype(int)

# AgeGroup — five life stages
df['AgeGroup'] = pd.cut(
    df['Age'],
    bins  = [0, 12, 18, 30, 50, 80],
    labels= ['Child','Teen','YoungAdult','Adult','Senior']
)

# FamilyCategory — three meaningful tiers
df['FamilyCategory'] = pd.cut(
    df['FamilySize'],
    bins  = [-1, 0, 3, 10],
    labels= ['Alone','Small Family','Large Family']
)

print('Engineered Features:')
print(f'  FamilySize    : {df["FamilySize"].min()}–{df["FamilySize"].max()} '
      f'(mean = {df["FamilySize"].mean():.2f})')
print(f'  IsAlone       : {df["IsAlone"].sum()} alone ({df["IsAlone"].mean()*100:.1f}%)')
print(f'  AgeGroup      : {df["AgeGroup"].value_counts().to_dict()}')
print(f'  FamilyCategory: {df["FamilyCategory"].value_counts().to_dict()}')
print()
print(f'Final shape: {df.shape} ({df.shape[1] - 12} features added)')
```

## Dashboard — Part I: Who Was Aboard?

```python
# ─────────────────────────────────────────────────────────────────────────────
# VIZ 11 — EMBARKATION PORT ANALYSIS
# Question: Did port of departure correlate with survival?
# ─────────────────────────────────────────────────────────────────────────────
EMB_COLORS = {'C':'#f472b6','Q':'#34d399','S':'#60a5fa'}
EMB_NAMES  = {'C':'Cherbourg (C)','Q':'Queenstown (Q)','S':'Southampton (S)'}

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# LEFT — survival rate
emb_rate = df.groupby('Embarked')['Survived'].mean() * 100
emb_n    = df.groupby('Embarked')['Survived'].count()
bs = axes[0].bar([EMB_NAMES[k] for k in emb_rate.index], emb_rate.values,
                  color=[EMB_COLORS[k] for k in emb_rate.index], edgecolor='none', width=0.5)
for b, v, n in zip(bs, emb_rate.values, emb_n.values):
    axes[0].text(b.get_x()+b.get_width()/2, b.get_height()+0.8,
                 f'{v:.1f}%\n(n={n})', ha='center', va='bottom', fontsize=9.5, fontweight='bold')
axes[0].axhline(df['Survived'].mean()*100, color='#fde68a', linestyle=':', lw=1.8,
                label=f'Overall = {df["Survived"].mean()*100:.1f}%')
axes[0].set_title('Survival Rate by Port')
axes[0].set_ylabel('Survival Rate (%)'); axes[0].set_ylim(0, 70); axes[0].legend()
axes[0].tick_params(axis='x', rotation=10)

# CENTRE — class composition per port
pct = df.groupby('Embarked')['Pclass'].value_counts(normalize=True).mul(100).unstack().fillna(0)
port_labels = [EMB_NAMES[k] for k in pct.index]
axes[1].bar(port_labels, pct[1], color=C_P1, edgecolor='none', label='1st Class')
axes[1].bar(port_labels, pct[2], color=C_P2, edgecolor='none', bottom=pct[1], label='2nd Class')
axes[1].bar(port_labels, pct[3], color=C_P3, edgecolor='none',
            bottom=pct[1]+pct[2], label='3rd Class')
axes[1].set_title('Class Composition by Port')
axes[1].set_ylabel('Percentage (%)'); axes[1].legend()
axes[1].tick_params(axis='x', rotation=10)

# RIGHT — KDE of fare by port
for emb in ['C','Q','S']:
    fares = df[df['Embarked']==emb]['Fare']
    fares[fares < 200].plot.kde(ax=axes[2], label=EMB_NAMES[emb],
                                 color=EMB_COLORS[emb], lw=2.5)
axes[2].set_title('Fare Distribution by Port (Fare < £200)')
axes[2].set_xlabel('Fare (£)'); axes[2].set_ylabel('Density')
axes[2].legend()

fig.suptitle('Viz 11 — Embarkation Port: Survival, Class & Fare', fontsize=14, fontweight='bold')
plt.tight_layout()
save('survival_by_embarkation.png')
plt.show()
```

## Conclusion

This dashboard has demonstrated that professional data visualization is not about producing the most charts — it is about producing the *right* charts, arranged in a narrative sequence that guides the reader from question to insight to implication.

**Objectives Achieved:** All five required chart types were delivered — histogram (age distribution), bar chart (gender/class/family survival), boxplot (fare by class), scatterplot (age vs fare), and correlation heatmap. Fifteen total visualisations were produced and saved as 300 DPI PNGs in the `images/` folder. Feature engineering created four meaningful variables: `FamilySize`, `IsAlone`, `AgeGroup`, and `FamilyCategory`.

**Visual Storytelling Outcomes:** The dashboard tells a coherent four-chapter story — the passenger profile, the survival patterns, the deep interactions, and the economic signals — culminating in a single-page executive summary suitable for stakeholder presentation. Every chart immediately answers a clearly stated question.

**Importance of Visualization:** The FamilySize finding is the clearest possible demonstration of why visualization is irreplaceable. A correlation of r = 0.017 declared no relationship; the bar chart revealed a decisive, non-linear inverted-U pattern that a machine learning model would need to discover on its own. Visualization surfaces this in seconds.

**Lessons Learned:** Chart selection matters more than chart quantity. A scatterplot, a violin plot, and a FacetGrid each reveal aspects of the data that a bar chart cannot. The choice of chart type is an analytical decision, not an aesthetic one.

**Future Improvements:** A deployed interactive version of this dashboard (using Plotly or Streamlit) would allow stakeholders to filter by class, gender, or age group dynamically. A predictive layer using logistic regression could quantify feature importance with confidence intervals, complementing the visual findings.
